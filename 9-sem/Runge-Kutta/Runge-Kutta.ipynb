# Runge-Kutta
## Task
Find a partial solution of the differential equation
```math
y^{'} + 2y = x^{2}
```
corresponding to the initial condition 
```math
y(0) = 1
```
using the Euler method on the segment [0; 1] with a step h = 0.1 . 

Build a table and graph of the approximate solution.

## Program

To calculate and visualize the results, we need the numpy and pyplot modules

````python
import numpy as np
import matplotlib.pyplot as plt
````

Define the function for the right-hand side of the equation:

````python
def f(x, y):
    return x**2 - 2*y
````

Euler method implementation:

````python
def euler_method(f, y0, x_start, x_end, h):
    # Number of steps
    steps = int((x_end - x_start) / h)
    
    # Arrays to store x and y
    xes = np.linspace(x_start, x_end, steps + 1)
    yes = np.zeros(steps + 1)
    
    yes[0] = y0

    for n in range(steps):
        yes[n + 1] = yes[n] + h * f(xes[n], yes[n])
    
    return xes, yes
````

Specify the parameters. You can change them if necessary.

````python
y0 = 1          # Initial condition
x_start = 0     # Start
x_end = 1       # End
h = 0.1         # Step
````

Run the Euler method:

````python
x_values, y_values = euler_method(f, y0, x_start, x_end, h)

print(f"{'x':>10} {'y':>10}")
print("-" * 20)
for x, y in zip(x_values, y_values):
    print(f"{x:10.2f} {y:10.6f}")
````

Plotting the results:

````python
plt.plot(x_values, y_values, label='Euler Method', marker='o')
plt.xlabel('x')
plt.ylabel('y')
plt.legend()
plt.grid()
plt.show()
````
